# 실제 사이트 - https://books.toscrape.com/

```python
import requests
from bs4 import BeautifulSoup 
url = 'https://books.toscrape.com/'
response = requests.get(url)
response.raise_for_status() # 에러체크

soup = BeautifulSoup(response.text, 'lxml')
# soup.select('article.product_pod')[0].find('h3').find('a').get('title') #get 더 안전해서 사용한다.
soup.select('article.product_pod')[0].select_one('h3 a').get('title')
```

```python
import requests
from bs4 import BeautifulSoup 
titles, links = [],[]
for pagenum in range(1,51):
    url = f'https://books.toscrape.com/catalogue/page-{pagenum}.html'
    response = requests.get(url)
    response.raise_for_status() # 에러체크

    soup = BeautifulSoup(response.text, 'lxml') # HTML 코드를 파이썬 객체로 변환  , 'lxml'은 **파서(parser)**의 종류로, 속도가 빠르고 HTML 구조 오류에도 잘 대응합니다.
    
    for li in soup.select('ol.row li'):
        # print(li.select('article.product_pod')[0].select_one('h3 a').get('title'))  # 크롤링
        titles.append(li.select('article.product_pod')[0].select_one('h3 a').get('title'))
        links.append('https://books.toscrape.com/'+li.select('article.product_pod')[0].select_one('h3 a').get('href'))
```

```python
import pandas as pd
data = {
    'title' : titles,
    'link' : links
}
pd.DataFrame(data)
```
