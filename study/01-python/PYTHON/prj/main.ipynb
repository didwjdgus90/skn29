```python
# 직원정보
# person = {
#    "id" : "sk-001",
#    "이름": "김철수",
#    "나이": 35,
#    "직급": "과장",
#    "월급": 5000,
#    "스킬": ["Python", "JavaScript", "SQL"],
#    "주소": {
#        "시": "서울",
#        "구": "강남구"
#    }
#}
# 입사 - 직원 dict 생성
# 조회
# 변경
# 삭제
# 직원 정보는 파일에 저장

# 요구사항
# 1. 생성, 조회, 변경, 삭제 종료 메뉴를 제공하고 사용자가 해당 메뉴를 선택하면 적절한 로직을 수행한다.
# 2. 프로그램 시작시 파일에서 직원정보를 load 한다.
# 3. 프로그램 종료전에 수정된 직원정보를 파일에 저장한다.
```

```python
# 직원정보는 어떻게 로드 : json 구조 -dictionary
# 생성 : input을 이용해서 사용자로부터 데이터를 직접 입력 6개
# 조회 : 사용자로부터 키값을 입력 받아서 출력 - 이름으로 조회하되 id정보는 반드시 보여줌
# 변경 : id값으로 찾아서 변경한다.
# 삭제 : 조회를 해서 나온 직원중에 삭제하고자 하는 직원의 id로 삭제한다.
# 종료 : 프로그램에서 사용한 모든 리소스를 해제하고 종료 - 파일 업데이트 (덮어쓰기)
```

```python
import json         # json을 다루기 위한 라이브러리 호출함

def load_data():    # load_data() 함수를 만들거임
    employees_file = 'data/emp.json' # employees_file 변수에 emp.json 파일의 경로를 저장함
    try:   # 오류가 있을시
        with open(employees_file,'r',encoding='utf-8') as f: # 해당 변수에 저장된 경로를 열어서 읽어오는 코드
            employee = json.load(f) #  로드 된 파일을 employees 변수에 저장
        return employee # employees를 반환한다.
    except FileNotFoundError as e: # 파일이 해당 경로 에 없을 시
        print(f'파일경로 확인해 주세요 : {e}') # 파일경로 확인 메시지를 출력한다.
    except Exception as e: # 이외의 에러가 발생시
        print(f'알수없는 오류 관리자 문의 : {e}') # 해당 메시지를 출력한다.

def save_data(employee): # employees를 받아서 save_data()함수를 만듬
    employees_file = 'data/emp.json' # employees_file 변수에 emp.json 파일의 경로를 저장함
    try: # 오류가 있을시
        with open(employees_file,'w',encoding='utf-8') as f: # 해당 변수에 저장된 경로를 열어서 쓰는 코드
            json.dump(employee, f, ensure_ascii=False, indent=2)  # employee 데이터를 보기 좋게 들여쓰기해서 Json 파일에 저장
    except FileNotFoundError as e: # 파일이 해당 경로에 없을 시
        print(f'파일경로 확인해 주세요 : {e}') # 해당 메시지 출력
    except Exception as e: # 이외의 에러 발생시
        print(f'알수없는 오류 관리자 문의 : {e}') # 해당 메시지 출력

# 조회 프로그램
def search_emp(employee): # employees를 받아서 search_emp() 함수 만듬
    keyword = input("검색할 이름 : ") # 조회하는 검색 창 만듬 keyword 변수에 저장
    found = True # found 변수에 참값을 저장
    for emp in employee: # employee에 있는 데이터 하나씩 추출해서 emp에 넣으면서 반복함
        if keyword in emp['name']: # 검색한 이름이 emp 이름에 있으면 
            for key,value in emp.items(): # 키 값과 속성 값들을 쌍으로 가져옴
                print(f"{key} : {value}") # 키 값과 속성 값 출력
        if not found: # 만약 찾지 못하면
            print("이름을 찾을수 없음") # 해당 메시지 출력
        else:  # 찾았으면
            print("-"*30) # 구분선 출력
    
# 아이디 인덱스
def insert_id(employee): # employee를 받아서 insert_id() 함수를 만듬
    if not employee: # 만약 employee 데이터가 없으면 
        return 0 # 0을 반환
    max_num = 0 # max_num 변수 초기값 설정
    for emp in employee: # employee 리스트를 하나씩 반복
        num = int(emp['id'].split("-")[1]) # id부분에서 숫자 부분만 추출해서 정수로 변환 해서 num에 저장
        if num > max_num : # 만약 max_num이 num보다 크면
            max_num = num # 갱신해준다.
    return f"sk-{max_num + 1:03d}" # sk- 다음번호로 이동하면서 + 1 세자리 채우기 ( 001 처럼 자리수 맞추기)

# 삽입
def insert_emp(employee): # employee를 받아서 insert_emp() 함수를 만듬
    emp = {} # 딕셔너리 형태 emp 만듬 
    emp['id'] = insert_id(employee) # 위에 만들었던 함수 insert_id를 직원 아이디로 사용한다.
    print(f'자동 생성된 아이디 : {emp["id"]}') # 자동 생성된 아이디여서 출력 메시지를 남긴다.
    emp['name'] = input("이름 입력 : ") # 직원 이름 입력
    emp['age'] = input("나이 입력 : ") # 직원 나이 입력
    emp['rank'] = input("직급 입력 : ") # 직원 직급 입력
    emp['money'] = input("월급 입력 : ") # 직원 월급 입력

    skills = input("스킬 입력(, 구분) : ").split(',') # 사용자가 입력한 스킬들을 , 기준으로 나누고 
    emp['skill'] = [s.strip() for s in skills] # 공백을 제거해서 직원 스킬에 저장한다.

    emp['addr'] = {
        "시" : input("시 : "),
        "구" : input("구 : ")
    } # 직원 주소는 시 랑 구로 입력 받는다.

    employee.append(emp) # employee에 입력받은 내용을 추가한다.
    print("직원 추가 완료") # 추가 내용 알림 메시지 출력


# 삭제
def delete_emp(employee): # employee 매개변수를 받아 delete_emp() 함수 만듬
    del_id = input("삭제할 아이디 : ") # 삭제할 아이디 검색

    for emp in employee: # employee 리스트를 하나씩 반복
        if emp['id'] == del_id: # 만약 검색한 아이디가 데이터에 있는 아이디와 일치하면
            employee.remove(emp) # 데이터 삭제
            print("삭제 완료") # 데이터 삭제 메시지 출력
            
# 수정
def update_emp(employee): # employee 매개변수를 받아 update_emp() 함수 만듬
    emp_id = input("수정할 아이디 검색 : ") # 수정할 아이디 검색

    for emp in employee: # employee 리스트를 하나씩 반복
        if emp['id'] == emp_id: # 만약 수정할 아이디랑 직원 아이디랑 같으면 

            name = input(f'이름 : {emp["name"]}') # 수정할 이름 입력
            if name: # 입력 이름 값이 있으면 실행
                emp['name'] = name # 직원이름에 입력된 이름으로 변경
                
            age = input(f'나이 : {emp["age"]}') # 수정할 나이 입력
            if age: # 입력 나이 값이 있으면 실행
                emp['age'] = age # 직원나이에 입력된 나이로 변경
            
            rank = input(f'직급 : {emp["rank"]}') # 수정할 직급 입력
            if rank: # 입력 직급 값이 있으면 실행
                emp['rank'] = rank # 직원 직급에 입력된 직급으로 변경
            
            money = input(f'월급 : {emp["money"]}') # 수정할 월급 입력
            if money: # 입력 월급 값이 있으면 실행
                emp['money'] = money # 직원 월급에 입력된 월급으로 변경
            
            print("수정완료") # 수정하였으면 메시지 출력
            return 0 # 0 반환
    print("ID 불일치") # ID가 일치하지 않을 때 메시지 출력


# 메뉴
def showMenu(): # 메뉴 함수 만듬
    employee = load_data() # 위의 json 파일 불러오는 함수 호출
    try: # 프로그램 실행중 에러나도 안죽고 처리
        while True: # 무한 반복 종료 될때까지
            choice = int(input("1. 생성 2. 조회 \n3. 수정 4. 삭제\n5. 종료")) # 번호 선택 항목 입력 
            if  1<= choice <= 5: # 1~ 5 사이 숫자만 처리
                if choice == 1: # 1번 누를시 
                    insert_emp(employee) # 직원 추가
                elif choice == 2: # 2번 누를시
                    search_emp(employee) # 직원 검색
                elif choice == 3: # 3번 누를시
                    update_emp(employee) # 직원 수정
                elif choice == 4: # 4번 누를시
                    delete_emp(employee) # 직원 삭제
                elif choice == 5: # 5번 누를시
                    save_data(employee) # 데이터 저장 후 종료
                    print("종료")
                    break
                else:
                    print("잘못된 입력") # 범위 잘못 입력 할 시
    except Exception as e: # 에러 메시지 출력
        print(f'올바른 번호를 선택하세요 {e}')

choice = showMenu() # 메뉴 입력
```
